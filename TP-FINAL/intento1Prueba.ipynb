```python
import tensorflow as tf
import cv2
import os
import numpy as np
from tensorflow.keras.utils import load_img, img_to_array

# Parámetros
IMAGE_SIZE = 150
pathDataset = "./dataset"

# Obtener nombres de clases
CLASS = sorted(os.listdir(pathDataset + "/train"))  # orden alfabético por consistencia
numClasses = len(CLASS)
print("Número de clases:", numClasses)

# Cargar modelo
bestModel = "./bestModel.keras"
model = tf.keras.models.load_model(bestModel)
model.summary()

# Preprocesar imagen
def preapImg(pathImg):
    img = load_img(pathImg, target_size=(IMAGE_SIZE, IMAGE_SIZE))
    img_array = img_to_array(img)
    img_array = np.expand_dims(img_array, axis=0)
    img_array = img_array / 150.0
    return img_array

# Ruta a la imagen de prueba
testImgPath = "./ImgPrueba1.png"

# Predecir
imForModel = preapImg(testImgPath)
predicciones = model.predict(imForModel, verbose=0)

indice_predicho = np.argmax(predicciones)
clase_predicha = CLASS[indice_predicho]

print(f"Índice predicho: {indice_predicho}")
print(f"Clase predicha: {clase_predicha}")

#-------------------------------------

# Ruta a la imagen de prueba
testImgPath = "./ImgPrueba2.png"

# Predecir
imForModel = preapImg(testImgPath)
predicciones = model.predict(imForModel, verbose=0)

indice_predicho = np.argmax(predicciones)
clase_predicha = CLASS[indice_predicho]

print(f"Índice predicho: {indice_predicho}")
print(f"Clase predicha: {clase_predicha}")
```
