```python
import tensorflow as tf
import os
import numpy as np
import cv2
from tensorflow.keras.utils import load_img, img_to_array
from IPython.display import display,HTML
from PIL import Image

IMAGE_SIZE = 150
pathDataset = "./dataset"

CLASS = sorted(os.listdir(pathDataset + "/train"))

bestModel = "./bestModel.keras"
model = tf.keras.models.load_model(bestModel)
model.summary()

def preapImg(pathImg):
    img = load_img(pathImg, target_size=(IMAGE_SIZE, IMAGE_SIZE))
    img_array = img_to_array(img)  # rango [0, 255]
    img_array = (img_array / 127.5) - 1  # igual que Rescaling
    img_array = np.expand_dims(img_array, axis=0)
    return img_array

pathImg = "./ImgPrueba1.png"
img = cv2.imread(pathImg)
imgForModel = preapImg(pathImg)
rstArray = model.predict(imgForModel, verbose=1)
ans = np.argmax(rstArray, axis=1)

print("La clase es " + CLASS[ans[0]])

```
