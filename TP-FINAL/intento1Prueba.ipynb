```python
import tensorflow as tf
import cv2
import os

from keras.utils import load_img, img_to_array
import numpy as np

IMAGE_SIZE = 224
pathDataset = "C:/Users/Ramiro/Desktop/Inteligencia Artigicial/TPs/IIA-TPS/TP-FINAL/dataset"
path = "C:/Users/Ramiro/Desktop/Inteligencia Artigicial/TPs/IIA-TPS/TP-FINAL"

CLASS = os.listdir(pathDataset + "/train")
numClasses = len(CLASS)
print("Numero de clases: " + str(numClasses))

bestModel = "C:/Users/Ramiro/Desktop/Inteligencia Artigicial/TPs/IIA-TPS/TP-FINAL/bestModel.keras"
model = tf.keras.models.load_model(bestModel)

print(model.summary())

def preapImg(pathImg):
    img = load_img(pathImg, target_size = (IMAGE_SIZE, IMAGE_SIZE))
    imgRst = img_to_array(img)
    imgRst = np.expand_dims(imgRst, axis = 0)
    imgRst = imgRst / 255
    return imgRst

# Selecciono imagen de prueba
testImgPath = path + "/ImgPrueba"

img = cv2.imread(testImgPath)
imgForModel = preapImg(testImgPath)
rstArray = model.predict(imgForModel, verbose = 1)

resultado = np.argmax(rstArray, axis = 1)
print(resultado)
```
