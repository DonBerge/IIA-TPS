```python
import os
import numpy as np
from tensorflow.keras.models import load_model
from tensorflow.keras.preprocessing import image

def classify_images_with_model(model_path, images_dir, target_size=(224, 224), class_names=None):
    """
    Clasifica imágenes en un directorio usando un modelo de Keras.

    Args:
        model_path (str): Ruta al archivo del modelo Keras (.h5).
        images_dir (str): Directorio con imágenes a clasificar.
        target_size (tuple): Tamaño al que redimensionar las imágenes.
        class_names (list): Lista opcional de nombres de clases.

    Returns:
        dict: Diccionario con nombre de archivo como clave y clase predicha como valor.
    """
    model = load_model(model_path)
    results = {}
    for fname in os.listdir(images_dir):
        fpath = os.path.join(images_dir, fname)
        if not os.path.isfile(fpath):
            continue
        try:
            img = image.load_img(fpath, target_size=target_size)
            x = image.img_to_array(img)
            x = np.expand_dims(x, axis=0)
            x = x / 255.0  # Normalización estándar
            preds = model.predict(x)
            pred_class = np.argmax(preds, axis=1)[0]
            if class_names:
                pred_label = class_names[pred_class]
            else:
                pred_label = str(pred_class)
            results[fname] = pred_label
        except Exception as e:
            results[fname] = f"Error: {e}"
    return results

results = classify_images_with_model(
    model_path='91porciento.keras',
    images_dir='prueba',
    target_size=(300, 300),
    class_names=['bird', 'monkey', 'boar', 'tiger', 'rat', 'ram', 'dog', 'horse', 'hare', 'ox', 'dragon', 'snake']  # Opcional
)

# Imprimir resultados
for fname, label in results.items():
    # Mostrar la imagen y su etiqueta
    print(f"{fname}: {label}")
```
