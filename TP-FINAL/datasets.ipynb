```python
from roboflow import Roboflow
import os
import shutil
rf = Roboflow(api_key="QgbgJ8ChSc0ml32j30r4")
project = rf.workspace("vgu-aeaes").project("naruto-hand-sign")
version = project.version(2)                
dataset = version.download("multiclass", location="dataset")

train_path = 'dataset/train'
validation_path = 'dataset/valid'
test_path = 'dataset/test'

def organize_images_by_first_word(base_path):
    for fname in os.listdir(base_path):
        fpath = os.path.join(base_path, fname)
        if os.path.isfile(fpath):
            word = fname.split('_')[0]
            target_dir = os.path.join(base_path, word)
            os.makedirs(target_dir, exist_ok=True)
            shutil.move(fpath, os.path.join(target_dir, fname))

organize_images_by_first_word(train_path)
organize_images_by_first_word(validation_path)
organize_images_by_first_word(test_path)

print("Path to dataset:", dataset.location)
```
