```python
import keras
from tensorflow.keras.preprocessing.image import (
    ImageDataGenerator
)

used_classes = ['bird', 'monkey', 'boar', 'tiger', 'rat', 'ram', 'dog', 'horse', 'hare', 'ox', 'dragon', 'snake']
IMAGE_SIZE = 300

BATCH_SIZE = 16

gen = ImageDataGenerator()

train_dataset = gen.flow_from_directory(
    'dataset/train',
    target_size=(IMAGE_SIZE, IMAGE_SIZE),
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    classes=used_classes,
    shuffle=True,
    seed=42
)

import matplotlib.pyplot as plt

class_counts = train_dataset.classes
# Invert class_indices to get a mapping from index to class name
index_to_class = {v: k for k, v in train_dataset.class_indices.items()}
class_labels = [index_to_class[i] for i in range(len(used_classes))]

counts_per_class = [sum(class_counts == i) for i in range(len(used_classes))]

plt.figure(figsize=(10, 6))
plt.bar(class_labels, counts_per_class)
plt.xlabel('Clase')
plt.ylabel('Cantidad de ejemplos')
plt.title('Cantidad de ejemplos por clase en el conjunto de entrenamiento')
plt.xticks(rotation=45)
plt.show()
```
